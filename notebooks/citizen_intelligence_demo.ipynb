## 1. Environment & Setup

```python
import sys
sys.path.append('..')
import pandas as pd
import numpy as np
from jan_sarthi_agent.agent import root_agent, process_feedback_intake, detect_hotspots, run_evidence_and_priority_pipeline, produce_policy_brief
print('JAN-SARTHI AI successfully loaded.')
```

## 2. Ingesting Unstructured Multilingual Feedback with PII Scrubber
Notice how phone numbers and private addresses are scrubbed, language is detected, and category is extracted.


```python
raw_hindi_input = 'हमारे गांव रामपुर में पिछले 20 दिनों से पानी की सप्लाई पूरी तरह बंद है। संपर्क करें: 9876543210 (रमेश, मकान नंबर 12).'
result = process_feedback_intake(
    text=raw_hindi_input,
    location_name='Rampur Rural',
    ward_id='WARD-01',
    latitude=28.74,
    longitude=77.14,
    reported_duration='20 days'
)
print('Status:', result['status'])
print('Detected Language:', result['feedback']['language'])
print('Extracted Category:', result['feedback']['category'])
print('Sanitized Text:', result['feedback']['normalized_text'])
print('PII Scrubbed:', result['security']['pii_entities_detected'])
```

## 3. Spatial-Temporal Hotspot Detection
Load baseline citizen signals and cluster them using DBSCAN.


```python
feedback_df = pd.read_csv('../data/demo_feedback.csv')
hotspots = detect_hotspots(feedback_df.to_dict(orient='records'))
print(f'Identified {len(hotspots)} active infrastructure hotspots:')
for h in hotspots:
    print(f"- {h['title']}: {h['raw_signal_count']} signals (Sector: {h['category']})")
```

## 4. Multi-Domain Evidence Fusion & Deterministic MCDA
Fuses citizen signals with demographic census vulnerability and municipal infrastructure baseline capacity.


```python
demographics_df = pd.read_csv('../data/demographics.csv')
infra_df = pd.read_csv('../data/infrastructure.csv')

target_cluster = hotspots[0]
pipeline_res = run_evidence_and_priority_pipeline(target_cluster, demographics_df, infra_df, feedback_df)
score = pipeline_res['priority_score']
print('Deterministic Decision-Support Priority Score:', score['composite_priority_index'])
print('Calculation Breakdown:', score['calculation_breakdown'])
```

## 5. Structured Policy Brief Generation
Generates evidence-backed policy brief strictly adhering to Zero-LLM-Arithmetic and Causality Guardrails.


```python
brief_res = produce_policy_brief(pipeline_res['evidence_packet'], score)
pb = brief_res['policy_brief']
print('EXECUTIVE PROBLEM STATEMENT:', pb['executive_problem_statement'])
print('RECOMMENDED INTERVENTION:', pb['recommended_capital_intervention'])
print('HUMAN VERIFICATION REQUIRED:', pb['human_verification_required'])
```
