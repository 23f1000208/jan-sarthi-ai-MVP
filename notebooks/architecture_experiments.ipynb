## 3. Post-LLM Deterministic Output Validation
Verifies that generated text cannot invent fictitious numbers or locations.


```python
from jan_sarthi_agent.security.validators import validate_policy_output

hallucinated_text = 'A major flood caused 8888 citizens to evacuate WARD-99.'
valid, violations = validate_policy_output(
    generated_text=hallucinated_text,
    evidence_numbers=[100.0, 20.0],
    allowed_locations=['WARD-01', 'WARD-02']
)
print('Output Valid:', valid)
print('Violations Intercepted:', violations)
```
